````python
# Create an API Client
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"

system_prompt = """
You are a Python coding assistant who write very consise code, the response should be fit in 50 token max only, retrun only the code without any explanation or text, and do not include any comments in the code.
"""
````

````python
def add_message(role, messages, text):
    messages.append({"role": role, "content": text})

# Make a request to the API
def chat(messages, system=None, stop_sequences=None):
    params = {
        "model": model,
        "max_tokens": 500,
        "messages": messages
    }

    if system:
        params["system"] = system

    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    message = client.messages.create(**params)    
    return message.content[0].text
````

````python
messages = []

add_message("user", messages, "write for me a python function that checks for duplicated caracters in string")
answer = chat(messages)
add_message("assistant", messages, answer)
answer

````
