````python
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()

client = Anthropic()
model = "claude-haiku-4-5"
````

````python
def add_message(role, messages, text):
    messages.append({"role": role, "content": text})

def chat(messages, system=None, stop_sequences=None):
    params = {
        "model": model,
        "max_tokens": 100,
        "messages": messages
    }

    if system:
        params["system"] = system

    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    message = client.messages.create(**params)    
    return message.content[0].text
````

````python
messages = []
add_message("user", messages, "write 3 different aws cli commands, each should be very shourt")
add_message("assistant", messages, "Here is all the 3 commands in one markdown code block with no comments:\n```bash")

answer = chat(messages, stop_sequences=["```"])
answer.strip()
````
