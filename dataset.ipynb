````python
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()

client = Anthropic()
model = "claude-haiku-4-5"
````

````python
def add_message(role, messages, text):
    messages.append({"role": role, "content": text})

def chat(messages, system=None, stop_sequences=None):
    params = {
        "model": model,
        "max_tokens": 500,
        "messages": messages
    }

    if system:
        params["system"] = system

    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    message = client.messages.create(**params)    
    return message.content[0].text
````

````python
messages = []

prompt = """
I wanna do a prompt evaluation workflow and i need a json file contains 5 differents tasks about coding in json, python and typescript to test the prompt with in this format:
```json
[
    {
        "task": "description 1"
    },
    {
        "task": "description 2"
    }
    ...
]
```
"""


add_message("user", messages, prompt)
add_message("assistant", messages, "Here is 5 tasks about coding in python, json and typescript: ```json")

dataset = chat(messages, stop_sequences=["```"])
dataset
````

````python
import json

with open("dataset.json", "w") as file:
    json.dump(json.loads(dataset.strip()), file, indent=2)
````

````python
def run_prompt(test_case):
    prompt = f"""
    Please solve the following task:
        {test_case["task"]}
    """

    messages = []
    add_message("user", messages, prompt)
    return chat(messages)
````
