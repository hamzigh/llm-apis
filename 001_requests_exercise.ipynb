```python
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()

messages = []
user_exit = False

client = Anthropic()

system_prompt = "You are a helpful assistant that responds in haiku form."

def add_msg(messages, role, text):
    messages.append({"role": role, "content": text})

def chat(messages):
    msg = client.messages.create(
        model="claude-haiku-4-5",
        max_tokens=50,
        messages=messages
    )
    return msg.content[0].text


while not user_exit:
    user_input = input("> ")
    print("> ", user_input)

    if user_input.lower() == "exit":
        user_exit = True
        print("Exiting the chat.")
    else:
        add_msg(messages, "user", user_input)
        answer = chat(messages)
        add_msg(messages, "assistant", answer)
        print("Assistant:", answer)
```
